```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}

df = pd.DataFrame(data)
```

```python
df
```

```python
 # used to calcuclate a categorical data eg- category store etc given in the data
```

```python
 # Group by Category and the sum of Sales
```

```python
df.groupby('Category')['Sales'].sum()
# group by create object 
```

```python
df.groupby('Store')['Sales'].sum()
```

```python
# Group by several columns in one statement
df.groupby('Store')[['Sales','Quantity']].sum()
```

```python
# Group by Categiory and store
```

```python
df.groupby(['Store','Category'])['Sales'].sum()
```

# Aggregation

```python
# it means incluing several values to find out one output eg - mean , median , std, min, max , count  etc
```

```python
df['Sales'].mean()
```

```python
# aggreagtion also have a fnction
# i.e .agg
```
