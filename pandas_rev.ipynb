# Selecting Rows

```python
data = {
    "Name": ["Rahul", "Aman", "Priya"],
    "Age": [21, 22, 20],
    "City": ["Lucknow", "Delhi", "Mumbai"],
    "Marks": [85, 90, 88],
    "Course": ["BTech", "MTech", "BCA"]
}

df = pd.DataFrame(data)
df
```

```python
df.loc[0]
# location
# depends on the lable
```

```python
df.loc[[0,1]]
# location
# depends on the lable
```

```python
df.iloc[2]
# indexed location
# independent of lable in terms of location
```

# Selecting subset of rows and columns

```python
df.loc[[0,1]][["City","Marks"]]
```

```python
df.loc[[1,2]][["Name","Age"]]
```

# conditional selection

```python
# <, >, == etc
```

```python
df.loc[df["Age"]>21]
```

```python
# for multiple conditions
```

```python
df.loc[(df["Age"]>21) & (df["Marks"] > 85
                        )]
```
