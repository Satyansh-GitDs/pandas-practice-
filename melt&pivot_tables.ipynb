# Pivot tables

```python
import numpy as np 
import pandas as pd
```

```python
df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})
```

```python
df
```

```python
# melt 
melted_data=df.melt(
    id_vars=["country","year"],
    value_vars=['sales','profit'],
    var_name="melted_variables",
    value_name="values"
)
```

```python
melted_data.pivot(
    index=['country','year'],
    columns="melted_variables",
    values='values'
)
```
