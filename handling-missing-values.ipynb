# Handling Missing Values 

```python
import numpy as np
import pandas as pd
```

```python
# Finding missing values

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df=pd.DataFrame(data)
df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
# Removing null values
```

```python
df
```

```python
df.dropna()
```

```python
df.dropna(thresh= 3)
# only drop nullvalues greater than 3
```

# filling missing values

```python
df.fillna(0)
```

```python
# columnwise filling of nullvalues
```

```python
values={
    "A":100,
    "B":200,
    "C":300,
    "D":400
}
```

```python
df.fillna(values)
```

```python
df.fillna(df.mean())
```
