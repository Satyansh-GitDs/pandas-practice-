# Merging joining and concatination

```python
# Merging two dataframes or tables
```

```python
import numpy as np
import pandas as pd
```

```python
employees = pd.DataFrame(
    {
        'employee_id':[1,2,3,4,5],
        'name':['john','Anna','Peter','Linda','Bob'],
        'deparetment':['HR','IT','Finance','IT','HR']
    }
)

#Data frame 2 : Salary information
salaries=pd.DataFrame(
    {
        'employee_id':[1,2,3,6,7],
        'salary':[60000,50000,80000,55000,65000],
        'bonus':[5000,6000,7000,8000,9000]
        
    }
)
```

```python
# employee
salaries
```

```python
pd.merge(employees,salaries)
```

```python
pd.merge(employees,salaries, on='employee_id')
```

```python
# now about 'how' attribute (total 4 sub values - inner,outer,left,right) --- inner by default which maeans it will merege on the basis of leftmost part ie employye id indcies in inner side here which is upto 1-3
# ie 1,2,3
pd.merge(employees,salaries, on='employee_id', how='inner')
```

```python
# if we want to excute on the basis of the outer means the actual column attribute name
pd.merge(employees,salaries, on='employee_id', how='outer')
```

```python
# using left value for how attribute
pd.merge(employees,salaries, on ="employee_id", how='left')
```

```python
pd.merge(employees,salaries , on ="employee_id", how='right')
# it means it will be trated eith respect to salaries data frame
```

# Concatination of 2 data frame 

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df2,df1])
```

```python
 # on the basis of rows 
pd.concat([df1,df2] , axis =1)
```

# Joining two data fames

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df2.join(df1)
```

```python
df1.join(df2, how= 'outer')
```
